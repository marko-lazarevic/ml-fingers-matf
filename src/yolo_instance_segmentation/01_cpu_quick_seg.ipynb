```python
import os
import sys
from pathlib import Path

from ultralytics import YOLO

RUNS_DIR = "runs"
EGOHANDS_YAML = "data/egohands_yolo/data.yaml"
DEVICE = "cpu"
```

```python
# train_dir = SRC_DIR / "instance_segmentation" / "egohands_yolo" / "images" / "train"
# if not train_dir.exists() or not any(train_dir.iterdir()):
#     raise FileNotFoundError(
#         "EgoHands dataset nije spreman. Prvo pokreni notebooks/00_data_preprocessing.ipynb"
#     )
DATA_YAML = EGOHANDS_YAML
# assert DATA_YAML.exists(), f"Nedostaje {DATA_YAML}"
# print(f"Dataset OK: {DATA_YAML}")
```

```python
RUN_NAME = "cpu_quick_seg"

model = YOLO("yolov8n-seg.pt")
model.train(
    task="segment",
    data=str(DATA_YAML),
    epochs=30,
    imgsz=640,
    batch=8,
    device=DEVICE,
    workers=8,
    cache=True,
    amp=False,
    plots=True,
    project="instance_segmentation/runs/segment",
    name=RUN_NAME,
    exist_ok=True,
)
```
